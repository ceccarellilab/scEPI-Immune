```python
from pathlib import Path
import pandas as pd
import numpy as np
import scipy.sparse as sp
import h5py
import anndata as ad
import matplotlib.pyplot as plt
import scanpy as sc
from sklearn.mixture import GaussianMixture
import numpy as np
```

```python
# ==============================
# Paths
# ==============================
XENIUM_ROOT = Path("/home/fceccarelli/home3/OT_simulation/code/Pozniac/Xenium/Xenium")
OUTPUT_ROOT = Path("/home/fceccarelli/home3/OT_simulation/code/Pozniac/Xenium_adatas/")
OUTPUT_ROOT.mkdir(exist_ok=True)

# ==============================
# Helper: read 10x-style HDF5
# ==============================
def read_cell_feature_matrix(h5_path):
    with h5py.File(h5_path, "r") as f:
        data = f["matrix/data"][:]
        indices = f["matrix/indices"][:]
        indptr = f["matrix/indptr"][:]
        shape = f["matrix/shape"][:]  # (n_genes, n_cells)

        # IMPORTANT: Xenium stores CSC, not CSR
        X = sp.csc_matrix((data, indices, indptr), shape=shape)

        genes = [g.decode() for g in f["matrix/features/name"][:]]
        barcodes = [b.decode() for b in f["matrix/barcodes"][:]]

    return X, genes, barcodes


# ==============================
# Main loop
# ==============================
adatas = {}

for sample_dir in sorted(XENIUM_ROOT.iterdir()):
    if not sample_dir.is_dir():
        continue

    sample = sample_dir.name
    print(f"Processing sample: {sample}")

    # ------------------------------
    # Read cells metadata
    # ------------------------------
    cells = pd.read_csv(sample_dir / "cells.csv.gz")
    cells = cells.set_index("cell_id")

    # ------------------------------
    # Read expression matrix
    # ------------------------------
    X, genes, barcodes = read_cell_feature_matrix(
        sample_dir / "cell_feature_matrix.h5"
    )

    # ------------------------------
    # Align cells to matrix
    # ------------------------------
    # barcodes == cell_id (order matters)
    cells = cells.loc[barcodes]

    # ------------------------------
    # Create AnnData (cells × genes)
    # ------------------------------
    adata = ad.AnnData(
        X=X.T,  # transpose → cells × genes
        obs=cells,
        var=pd.DataFrame(index=genes)
    )

    # ------------------------------
    # Add spatial coordinates
    # ------------------------------
    adata.obsm["spatial"] = np.vstack([
        adata.obs["x_centroid"].values,
        adata.obs["y_centroid"].values
    ]).T

    # ------------------------------
    # Optional: transcripts
    # ------------------------------
    tx_path = sample_dir / "transcripts.csv.gz"
    if tx_path.exists():
        adata.uns["transcripts"] = pd.read_csv(tx_path)

    # ------------------------------
    # Metadata
    # ------------------------------
    adata.uns["sample_id"] = sample
    adata.uns["technology"] = "10x Xenium"

    var = adata.var.copy()

    var["is_neg_control"] = var.index.str.startswith("NegControlCodeword")
    var["is_unassigned"] = var.index.str.startswith("UnassignedCodeword")
    var["is_deprecated"] = var.index.str.startswith("DeprecatedCodeword")
    var["is_control"] = var.index.str.contains("Control", regex=True)

    var["is_gene"] = ~(
        var["is_neg_control"] |
        var["is_unassigned"] |
        var["is_deprecated"] |
        var["is_control"]
    )

    adata.var = var

    adata = adata[:, adata.var["is_gene"]].copy()

    adata.var["gene"] = True

    sc.pp.calculate_qc_metrics(
        adata,
        qc_vars=["gene"],
        percent_top=None,
        log1p=False,
        inplace=True
    )

    adata.obs["control_frac"] = (
        adata.obs["control_codeword_counts"] /
        adata.obs["total_counts"].clip(lower=1)
    )

    sc.pp.filter_cells(adata, min_counts=30)
    sc.pp.filter_cells(adata, min_genes=10)

    adata = adata[adata.obs["cell_area"] >= 30].copy()
    adata = adata[adata.obs["control_frac"] <= 0.2].copy()

    print (adata)
    print (adata.var_names)

    xy = adata.obsm["spatial"]

    plt.figure(figsize=(6, 6))
    plt.scatter(
        xy[:, 0],
        xy[:, 1],
        s=1,
        alpha=0.6
    )
    plt.gca().invert_yaxis()
    plt.title(f"Spatial cells – {sample}")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.tight_layout()
    plt.show()

    # ------------------------------
    # Save AnnData
    # ------------------------------
    out_file = OUTPUT_ROOT / f"{sample}.h5ad"
    adata.write(out_file)

    adatas[sample] = adata
    print(f"Saved → {out_file}\n")

print("All samples processed.")
```

```python
melanoma_genes = [
    'AXL', 'CD274', 'DCT', 'MITF', 'MKI67', 'MLANA',
    'NGFR', 'PMEL', 'S100A1', 'S100B', 'SOX10',
    'TOP2A', 'TYR', 'TYRP1'
]

import scanpy as sc
from pathlib import Path
import matplotlib.pyplot as plt

ADATA_ROOT = Path("/home/fceccarelli/home3/OT_simulation/code/Pozniac/Xenium_adatas")
SCORED_ROOT = ADATA_ROOT / "scored"
SCORED_ROOT.mkdir(exist_ok=True)

scored_adatas = {}

for adata_path in sorted(ADATA_ROOT.glob("*.h5ad")):
    sample = adata_path.stem
    print(f"Processing {sample}")

    adata = sc.read_h5ad(adata_path)

    # ------------------------------
    # Normalization (Xenium-appropriate)
    # ------------------------------
    sc.pp.normalize_total(
        adata,
        target_sum=1e4
    )
    sc.pp.log1p(adata)

    # ------------------------------
    # Melanoma gene scoring
    # ------------------------------
    present = [g for g in melanoma_genes if g in adata.var_names]

    sc.tl.score_genes(
        adata,
        gene_list=present,
        score_name="melanoma_score",
        use_raw=False
    )

    scores = adata.obs["melanoma_score"].values.reshape(-1, 1)

    gmm = GaussianMixture(
        n_components=2,
        random_state=0
    )
    gmm.fit(scores)

    # identify which component is tumor (higher mean)
    means = gmm.means_.flatten()
    tumor_component = np.argmax(means)

    # assign labels and probabilities
    adata.obs["tumor"] = (gmm.predict(scores) == tumor_component)
    adata.obs["tumor_prob"] = gmm.predict_proba(scores)[:, tumor_component]

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    # 1️⃣ Melanoma score
    sc.pl.spatial(
        adata,
        color="melanoma_score",
        size=1.2,
        spot_size=50,
        cmap="viridis",
        title="Melanoma score",
        ax=axes[0],
        show=False
    )

    # 2️⃣ Tumor probability (GMM)
    sc.pl.spatial(
        adata,
        color="tumor_prob",
        size=1.2,
        spot_size=50,
        cmap="viridis",
        title="Tumor probability (GMM)",
        ax=axes[1],
        show=False
    )

    # 3️⃣ Tumor vs non-tumor (binary)
    sc.pl.spatial(
        adata,
        color="tumor",
        size=1.2,
        spot_size=50,
        title="Tumor vs non-tumor (GMM)",
        ax=axes[2],
        show=False
    )

    plt.tight_layout()
    plt.show()

    scored_adatas[sample] = adata
    
    adata.uns["melanoma_genes_used"] = present
    adata.uns["gmm_means"] = means.tolist()

    out_path = SCORED_ROOT / f"{sample}_melanoma_scored.h5ad"
    adata.write(out_path)
```
