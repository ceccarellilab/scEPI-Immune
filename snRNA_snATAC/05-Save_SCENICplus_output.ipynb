# Downstream analysis
1. Direct gene-based AUC and direct region-based AUC of the regulons with a R2G corr > 0 

```python
gene_sig_name = filtered_df.Gene_signature_name.unique()
gene_sig_name = gene_sig_name.tolist()
```

```python
region_sig_name = filtered_df.Region_signature_name.unique()
region_sig_name = region_sig_name.tolist()
region_sig_name
```

```python
scplus_mdata['direct_gene_based_AUC'].var_names.to_list()
```

```python
selected_vars = gene_sig_name + region_sig_name
selected_vars
```

```python
gene_direct_auc = scplus_mdata.mod['direct_gene_based_AUC']

# Subset the modality to only include selected variables
scplus_mdata.mod['direct_gene_based_AUC'] = gene_direct_auc[:, gene_sig_name]
```

```python
scplus_mdata
```

```python
region_direct_auc = scplus_mdata.mod['direct_region_based_AUC']

# Subset the modality to only include selected variables
scplus_mdata.mod['direct_region_based_AUC'] = region_direct_auc[:, region_sig_name]
```

```python
scplus_mdata
```

```python
scplus_mdata.mod.keys()
```

```python
from scenicplus.regulon_qc.quality_metrics import generate_pseudobulks, calculate_correlation
gene_pseudobulk = generate_pseudobulks(scplus_mdata, 
                                       variable = 'scRNA_counts:Metaprogram_assignment',
                                       modality = 'direct_gene_based_AUC',
                                       nr_cells_to_sample = 10,
                                       nr_pseudobulks_to_generate = 100,
                                       seed=555
                                       )
```

```python
gene_pseudobulk
```

```python
region_pseudobulk = generate_pseudobulks(scplus_mdata, 
                                         variable = 'scRNA_counts:Metaprogram_assignment',
                                         modality = 'direct_region_based_AUC',
                                         nr_cells_to_sample = 10,
                                         nr_pseudobulks_to_generate = 100,
                                         seed=555
                                         )
```

```python
region_pseudobulk
```

```python
my_mapping_A_to_B = dict(zip(gene_sig_name, region_sig_name))
my_mapping_A_to_B
```

```python
corr_gene_region_AUC = calculate_correlation(gene_pseudobulk, region_pseudobulk, my_mapping_A_to_B)
```

```python
corr_gene_region_AUC
```

```python
robust_tf_gene = corr_gene_region_AUC.A[corr_gene_region_AUC.rho > 0.3].tolist()+corr_gene_region_AUC.A[corr_gene_region_AUC.rho < -0.3].tolist()
robust_tf_region = corr_gene_region_AUC.B[corr_gene_region_AUC.rho > 0.3].tolist()+corr_gene_region_AUC.B[corr_gene_region_AUC.rho < -0.3].tolist()
```

```python
scplus_mdata
```

```python
gene_direct_auc = scplus_mdata.mod['direct_gene_based_AUC']

# Subset the modality to only include selected variables
scplus_mdata.mod['direct_gene_based_AUC'] = gene_direct_auc[:, robust_tf_gene]
```

```python
scplus_mdata
```

```python
region_direct_auc = scplus_mdata.mod['direct_region_based_AUC']

# Subset the modality to only include selected variables
scplus_mdata.mod['direct_region_based_AUC'] = region_direct_auc[:, robust_tf_region]
```

```python
scplus_mdata
```
